```python
import pandas as pd
df = pd.read_csv("student_performance_5000.csv")
print(df.head())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
import matplotlib.pyplot as plt
import numpy as np


X = df[['Hours_Read']]
y = df['Marks']

#train model
model = LinearRegression()
model.fit(X,y)
predicted_score = model.predict(X)

mae = mean_absolute_error(y,predicted_score)
mse = mean_squared_error(y,predicted_score)
rmse = np.sqrt(mse)
r2 = r2_score(y,predicted_score)
print(" mean_absolute_error = ",round(mae,2)) # used round to round up to 2 decimal places
print(" mean_squared_error = ",round(mse,2))
print(" root_mean_absolute_error = ",round(rmse,2))
print(" r2 score(Model accuracy) = ",round(r2,2)) # r2 score is near 1 then better

# Plotting a histogram to show no. of students and marks
plt.figure(figsize = (10,6))
plt.hist(df["Marks"] , bins = 30 , color ="yellow" , edgecolor="black")
plt.title("Final marks distribution")
plt.xlabel("Marks")
plt.ylabel("No. of students")
plt.show()

# Plotting a scatter plot and regression line
plt.figure(figsize = (10,6))
plt.scatter(X,y ,color ="blue" , label="Actual Scores")
plt.plot(X, predicted_score, color="red" , label="Predicted Scores")
plt.title("Model prediction vs Actual Scores")
plt.xlabel("Study hours daily")
plt.ylabel("Final score")
plt.show()
```
