```python
# Handling datasets
import pandas as pd

data = {
    'Name' : [ 'Ram' , 'Shyam' , 'Raju' , 'Sonu' , 'Vivek'],
    'Age': [22 ,None , 44 ,  23 , None],
    'Salary' : [50000 , 60000 , 70000 , None , None]
}

df = pd.DataFrame(data)
print(df)
print("\nThe number of missing values in each column :=")
print(df.isnull().sum())

# Deleting NaN rows
print("\nAfter deleting NaN data rows")
df_drop = df.dropna()
print(df_drop)

# Replacing NaN values with mean of that column
print("\nReplacing NaN values with mean of that column")
df['Age'].fillna( df['Age'].mean() , inplace = True)
df['Salary'].fillna( df['Salary'].mean() , inplace = True)
print(df)
```

```python
# Label Encoding  ==> technique that converts categorical (text) values into numeric labels (integers) by assigning a unique number to each category.
from sklearn.preprocessing import LabelEncoder
import pandas as pd

data = {
    "Name": [
        "Ram", "Shyam", "Raju", "Sonu", "Vivek",
        "Priya", "Neha", "Anjali", "Pooja", "Ritika",
        "Aman", "Rohit", "Karan", "Simran", "Kavita",
        "Nisha", "Arjun", "Sneha", "Rahul", "Meera"
    ],
    "Gender": [
        "Male", "Male", "Male", "Male", "Male",
        "Female", "Female", "Female", "Female", "Female",
        "Male", "Male", "Male", "Female", "Female",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Passed": [
        "Yes", "No", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

df_new = df.copy()

le = LabelEncoder()

df_new['Gender_Encoded'] = le.fit_transform(df_new['Gender']) # assigns 1 to male and 0 to female
df_new['Passes_Encoded'] = le.fit_transform(df_new['Passed']) # assigns 1 to yes and 0 to no

print(df_new)
```

```python
# one hot encoding ==> technique that converts a categorical column into multiple binary (0 or 1) columns, where each unique category gets its own column.


import pandas as pd

data = {
    "City": [
        "Delhi", "Mumbai", "Kolkata", "Delhi", "Mumbai",
        "Kolkata", "Delhi", "Mumbai", "Kolkata", "Delhi",
        "Mumbai", "Kolkata", "Delhi", "Mumbai", "Kolkata",
        "Delhi", "Mumbai", "Kolkata", "Delhi", "Mumbai"
    ]
}

df = pd.DataFrame(data)

df_new = pd.get_dummies(df , columns = ['City'] , dtype=int)
print(df_new[['City_Delhi' , 'City_Mumbai' , 'City_Kolkata']])
```

```python
# Feature Scaling - Feature Scaling is the process of bringing numerical features into a similar range or scale so that
#                   no feature dominates the others because of its larger values.
# 1.Standard Scaler ==>  Converts values so that they have approximately:
#                        Mean = 0
#                        Standard deviation = 1
# Formula = (Actual value - Mean)/Standard Deviation

import pandas as pd
from sklearn.preprocessing import StandardScaler

data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 50000, 70000, 100000]
}

df = pd.DataFrame(data)

scaler = StandardScaler()

df_scaled = scaler.fit_transform(df)

df_scaled = pd.DataFrame(df_scaled, columns=df.columns)

print(df_scaled)
```

```python
# Feature Scaling
# 2.Min Max Scaler - converts values to a range of 0 to 1.

# Formula =>  (x') = (x - xmin)/(xmax - xmin)

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 50000, 70000, 100000]
}

df = pd.DataFrame(data)

scaler = MinMaxScaler()

df_scaled = scaler.fit_transform(df)

df_scaled = pd.DataFrame(df_scaled, columns=df.columns)

print(df_scaled)
```

```python
# Linear Regression - genrerally used to predict numbers
#Marks predictor

from sklearn.linear_model import LinearRegression

X = [[1] , [2] , [3] , [4] , [5]]
y = [ 60 , 65 , 70 , 75 , 90]

model = LinearRegression()
model.fit(X,y)

hours = float(input("Enter the number of hours you studied = "))
predict_marks = model.predict([[hours]]) # it will return a Numpy array
print(f"If you study {hours} hours you may score {predict_marks[0]:.2f}") # To print Numpy array as a floating value
```

```python
# Logistic Regression - generally used to predict binary values

# Pass of fail Predictor
from sklearn.linear_model import LogisticRegression

X = [[1] , [2] , [3] , [4] , [5]]
y = [ 0,0,1,1,1]

model = LogisticRegression()
model.fit(X,y)

hours = float(input("Enter the number of hours you studied = "))
result = model.predict([[hours]])

if result == 1:
  print(f"If you study {hours} hours you will be pass")
else:
  print(f"If you study {hours} hours you will be fail")
```

```python
# KNN - K nearest neighbors
# Fruit Predictor by its weight and size

from sklearn.neighbors import KNeighborsClassifier

X = [
    [180,7],
    [190,7.2],
    [200,7.5],
    [250,8],
    [300,8.5],
    [330,9],
    [360,9.5],
]

y = [0 , 0 , 0 , 1 , 1 , 1 , 1]

model = KNeighborsClassifier(n_neighbors = 3)
model.fit(X,y)

weight = float(input("Enter the weight of fruit in grams = "))
size = float(input("Enter the size of the fruit in cm = "))

prediction = model.predict([[weight , size]])
if prediction == 0 :
  print(f"The fruit of weight {weight} grams and size {size} cm is most likely Apple")
else:
  print(f"The fruit of weight {weight} grams and size {size} cm is most likely Orange")
```

```python
# Decision  tree
# Fruit predictor by using its size and color - I have prioritize color more than weight

from sklearn.tree import DecisionTreeClassifier

X = [
    [180, 1],  # Apple
    [190, 2],  # Apple
    [200, 3],  # Apple

    [210, 7],  # Orange
    [220, 8],  # Orange
    [230, 9],  # Orange

    # Additional examples showing color is more important
    [250, 1],  # Apple
    [240, 2],  # Apple
    [230, 3],  # Apple

    [180, 7],  # Orange
    [190, 8],  # Orange
    [200, 9],  # Orange
]

y = [
    0, 0, 0,
    1, 1, 1,
    0, 0, 0,
    1, 1, 1
]

model = DecisionTreeClassifier()
model.fit(X,y)

weight = float(input("Enter the weight of fruit in grams = "))
color_grade = float(input("Enter the color grade of the fruit in 0-10 = "))

prediction = model.predict([[weight , color_grade]])
if prediction == 0 :
  print(f"The fruit of weight {weight} grams and color grade {color_grade} is most likely Apple")
else:
  print(f"The fruit of weight {weight} grams and color grade {color_grade} is most likely Orange")
```

```python
# Evaluation metrics
# Accuracy , Precision , Recall , F1 score

from sklearn.metrics import accuracy_score , precision_score , recall_score , f1_score

x = [1,0,1,1,0,1,0] # actual data
y = [1,0,1,0,0,1,1] # predicted data

print("Accuracy  = " , accuracy_score(x,y)*100 ,"%")
print("Precision = " , precision_score(x,y)*100 ,"%")
print("Recall    = " , recall_score(x,y)*100 ,"%")
print("F1 score  = " , f1_score(x,y)*100 ,"%")
```

```python
#Confusion matrix
# output is in form of 2D matrix

#  [
#   [TN  FP]
#   [FN  TP]
#  ]

# TP (True Positive): The model correctly predicts the positive class.
# TN (True Negative): The model correctly predicts the negative class.
# FP (False Positive): The model incorrectly predicts the positive class when the actual class is negative.
# FN (False Negative): The model incorrectly predicts the negative class when the actual class is positive.

from sklearn.metrics import confusion_matrix

x = [1,0,1,1,0,1,0] # actual data
y = [1,0,1,0,0,1,1] # predicted

cm = confusion_matrix(x,y)

print(cm)


```
