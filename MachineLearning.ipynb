```python
# Handling datasets
import pandas as pd

data = {
    'Name' : [ 'Ram' , 'Shyam' , 'Raju' , 'Sonu' , 'Vivek'],
    'Age': [22 ,None , 44 ,  23 , None],
    'Salary' : [50000 , 60000 , 70000 , None , None]
}

df = pd.DataFrame(data)
print(df)
print("\nThe number of missing values in each column :=")
print(df.isnull().sum())

# Deleting NaN rows
print("\nAfter deleting NaN data rows")
df_drop = df.dropna()
print(df_drop)

# Replacing NaN values with mean of that column
print("\nReplacing NaN values with mean of that column")
df['Age'].fillna( df['Age'].mean() , inplace = True)
df['Salary'].fillna( df['Salary'].mean() , inplace = True)
print(df)
```

```python
# Label Encoding  ==> technique that converts categorical (text) values into numeric labels (integers) by assigning a unique number to each category.
from sklearn.preprocessing import LabelEncoder
import pandas as pd

data = {
    "Name": [
        "Ram", "Shyam", "Raju", "Sonu", "Vivek",
        "Priya", "Neha", "Anjali", "Pooja", "Ritika",
        "Aman", "Rohit", "Karan", "Simran", "Kavita",
        "Nisha", "Arjun", "Sneha", "Rahul", "Meera"
    ],
    "Gender": [
        "Male", "Male", "Male", "Male", "Male",
        "Female", "Female", "Female", "Female", "Female",
        "Male", "Male", "Male", "Female", "Female",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Passed": [
        "Yes", "No", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

df_new = df.copy()

le = LabelEncoder()

df_new['Gender_Encoded'] = le.fit_transform(df_new['Gender']) # assigns 1 to male and 0 to female
df_new['Passes_Encoded'] = le.fit_transform(df_new['Passed']) # assigns 1 to yes and 0 to no

print(df_new)
```

```python
# one hot encoding ==> technique that converts a categorical column into multiple binary (0 or 1) columns, where each unique category gets its own column.


import pandas as pd

data = {
    "City": [
        "Delhi", "Mumbai", "Kolkata", "Delhi", "Mumbai",
        "Kolkata", "Delhi", "Mumbai", "Kolkata", "Delhi",
        "Mumbai", "Kolkata", "Delhi", "Mumbai", "Kolkata",
        "Delhi", "Mumbai", "Kolkata", "Delhi", "Mumbai"
    ]
}

df = pd.DataFrame(data)

df_new = pd.get_dummies(df , columns = ['City'] , dtype=int)
print(df_new[['City_Delhi' , 'City_Mumbai' , 'City_Kolkata']])
```

```python
# Feature Scaling - Feature Scaling is the process of bringing numerical features into a similar range or scale so that
#                   no feature dominates the others because of its larger values.
# 1.Standard Scaler ==>  Converts values so that they have approximately:
#                        Mean = 0
#                        Standard deviation = 1
# Formula = (Actual value - Mean)/Standard Deviation

import pandas as pd
from sklearn.preprocessing import StandardScaler

data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 50000, 70000, 100000]
}

df = pd.DataFrame(data)

scaler = StandardScaler()

df_scaled = scaler.fit_transform(df)

df_scaled = pd.DataFrame(df_scaled, columns=df.columns)

print(df_scaled)
```

```python
# Feature Scaling
# 2.Min Max Scaler - converts values to a range of 0 to 1.

# Formula =>  (x') = (x - xmin)/(xmax - xmin)

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 50000, 70000, 100000]
}

df = pd.DataFrame(data)

scaler = MinMaxScaler()

df_scaled = scaler.fit_transform(df)

df_scaled = pd.DataFrame(df_scaled, columns=df.columns)

print(df_scaled)
```

```python
# Linear Regression - genrerally used to predict numbers
#Marks predictor

from sklearn.linear_model import LinearRegression

X = [[1] , [2] , [3] , [4] , [5]]
y = [ 60 , 65 , 70 , 75 , 90]

model = LinearRegression()
model.fit(X,y)

hours = float(input("Enter the number of hours you studied = "))
predict_marks = model.predict([[hours]]) # it will return a Numpy array
print(f"If you study {hours} hours you may score {predict_marks[0]:.2f}") # To print Numpy array as a floating value
```

```python
# Logistic Regression - generally used to predict binary values

# Pass of fail Predictor
from sklearn.linear_model import LogisticRegression

X = [[1] , [2] , [3] , [4] , [5]]
y = [ 0,0,1,1,1]

model = LogisticRegression()
model.fit(X,y)

hours = float(input("Enter the number of hours you studied = "))
result = model.predict([[hours]])

if result == 1:
  print(f"If you study {hours} hours you will be pass")
else:
  print(f"If you study {hours} hours you will be fail")
```
